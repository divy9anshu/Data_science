```python

import numpy as np

def CLT(data, no_of_samples, sample_size):
    result = []

    for i in range(no_of_samples):
        samples = np.random.choice(data, size=sample_size)
        sample_mean = np.mean(samples)
        result.append(sample_mean)
    result=np.array(result)
    return result
```

```python
no_of_samples=1000
smaple_size=30
CLT(price,no_of_samples,smaple_size)
```

```python
x=np.array([1,2,3,4,5])
y=2*x+1
plt.xlabel("x_values")
plt.ylabel("squares_of_x")
plt.title("Graph between x vs y")
plt.plot(x,y)
plt.show()
```
