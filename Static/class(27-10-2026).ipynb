```python
setosa=df[df["variety"]=="Setosa"]
versicolort=df[df["variety"]=="Versicolor"]
virginica=df[df["variety"]=="Virginica"]
```

```python
def outliers_range(data):
    Q1=np.percentile(data,25) 
    Q3=np.percentile(data,75) 
    IQR=Q3-Q1 
    LL=Q1-(1.5*IQR) 
    UL=Q3+(1.5*IQR) 
    return LL,UL 
```

```python
data=df["sepal.length"]
LL,UL=outliers_range(data)
```
