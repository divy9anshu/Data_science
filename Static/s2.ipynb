```python
import numpy as np
def chebyshevs_inequality(data,k):
    mean=np.mean(data)
    sd=np.std(data)
    LL=(mean-k*sd)
    UL=(mean+k*sd)
    per=(1-(1/k**2))*100
    return (LL,UL),per
```

```python
data=np.random.uniform(0,1,100)
k=2
chebyshevs_inequality(data,2)
```

```python
def central_limit_therom(data,no_of_samples,sample_size):
    result=[]
    for i in range(no_of_samples):
        sample=np.random.choice(data,size=sample_size,replace=True)
        result.append(np.mean(sample))
    data=np.array(result)
    return data
```

```python
data=np.random.randint(1,1000,1000)
no_of_samples=1000
sample_size=30
central_limit_therom(data,no_of_samples,sample_size)
```

```python

import numpy as np
import math

def gaussian_CDF(data):
    mean = np.mean(data)
    sd = np.std(data)
    result = []

    for i in data:
        ans = (i - mean) / (np.sqrt(2) * sd)
        res = 0.5 * (1 + math.erf(ans))
        result.append(res)

    return np.array(result)
```

```python
data3=np.sort(data)
gaussian_CDF(data3)
```
