```python
def central_limit_therom(data,no_of_samples,sample_size):
    result=[]
    for i in range(no_of_samples):
        sample=np.random.choice(data,size=sample_size,replace=True)
        result.append(np.mean(sample))
    data=np.array(result)
    return data
```

```python
data=df["Salary"]
no_of_samples=1000
sample_size=30
value=central_limit_therom(data,no_of_samples,sample_size)
```

```python
import math
def gaussian_CDF(data):
    mean = np.mean(data)
    sd = np.std(data)
    result = []

    for i in data:
        ans = (i - mean) / (np.sqrt(2) * sd)
        res = 0.5 * (1 + math.erf(ans))
        result.append(res)

    return np.array(result)
```

```python
def empirical_properties_GD(data,k):
    mean=np.mean(data)
    sd=np.std(data)
    LL=(mean-k*sd)
    UL=(mean+k*sd)
    return (LL,UL)
```

```python
rang = empirical_properties_GD(df["Salary"],2)
```

```python
def chebyshevs_inequality(data,k):
    mean=np.mean(data)
    sd=np.std(data)
    LL=(mean-k*sd)
    UL=(mean+k*sd)
    per=(1-(1/k**2))*100
    return (LL,UL),per
```
