```python
def central_limit_therom(data,no_of_samples,sample_size):
    result=[]
    for i in range(no_of_samples):
        sample=np.random.choice(data,size=sample_size,replace=True)
        result.append(np.mean(sample))
    data=np.array(result)
    return data
```

```python
data=df["sepal.length"]
no_of_samples=1000
sample_size=30
result=central_limit_therom(data,no_of_samples,sample_size)
```

```python
sns.distplot(result)
```

```python
def gaussian_CDF(data):
    mean = np.mean(data)
    sd = np.std(data)
    result = []

    for i in data:
        ans = (i - mean) / (np.sqrt(2) * sd)
        res = 0.5 * (1 + math.erf(ans))
        result.append(res)

    return np.array(result)
```

```python
data=df["sepal.length"]
gaussian_CDF(data)

sns.distplot(result)
```
