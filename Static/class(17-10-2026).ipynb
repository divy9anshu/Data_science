```python
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv("laptop_data.csv")
```

```python
sns.distplot(df["Price"])
```

```python
import numpy as np
import math
def Gaussian_CDF(data):
    result=[]
    meu=np.mean(data)
    sd=np.std(data)
    for i in data:
        a1=(i-meu)/(np.sqrt(2)*sd) 
        ans=1/2*(1+math.erf(a1)) 
        result.append(ans) 
    return np.array(result)       
```

```python
data=np.sort(df["Price"])
cdf_values=Gaussian_CDF(data)
```

```python
cdf_values
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.xlabel("x_values")
plt.ylabel("CDF values") 
plt.title("CDF")
plt.plot(data,cdf_values)
plt.show()
```
