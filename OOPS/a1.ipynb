```python
# implement a stack
class stack_implementation:
    def __init__(self):
        self.stack=[]
    def push(self,element):
        self.stack.append(element)
        return self.stack
    def pop(self):
        if self.isempty():
            return "stack is empty we cannot pop"
        else:
            self.stack.pop(-1)
            return self.stack
    def isempty(self):
        if len(self.stack)==0:
            return True
        else:
            return False
    def peep(self):
        return self.stack[-1],self.stack
```

```python
s=stack_implementation()
s.push(10)
```
