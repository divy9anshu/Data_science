# WAP a program to get the sum of all even number of n number

```python
n = int(input("Entr any number"))

i=1
even_sum=0
while i<=n:
    if i%2==0:
        even_sum+=i
    i+=1
print(even_sum)

        
```

# WAP to get difference betwn sum of even and sum of odd

```python
n = int(input("Entr any number"))
i=1
even_sum=0
odd_sum=0
while i<=n:
    if i%2==0:
        even_sum+=i
    else:
        odd_sum+=i
    i+=1
print(even_sum-odd_sum)
```

# WAP to print a pythoagorous triplets.

```python
n= int(input("Enter any number"))
i=1
while i<=n:
    j=i+1
    while j<=n:
        k=i+2
        while k<=n:
            if (i**2==j**2+k**2) or (j**2==i**2+k**2) or (k**2==i**2+j**2):
                print(i,j,k)
            k+=1
        j+=1
    i+=1
    
```

# WAP to get a sum of digits in a number :


```python
n= int(input("Enter the digit"))

sum=0
while n>0:
    digit=n%10
    sum+=digit
    n//=10
print(sum)
```
