```python
def convert_lst_dict(lst):
    new_dict={}
    for i in range(len(lst)):
        new_dict[lst[i]]=i
    return new_dict
```

```python
lst=[4,5,6,7,8]
convert_lst_dict(lst)
```

```python
def common_ele(lst1,lst2):
    dic={}
    result={}
    for i in lst2:
            dic[i]=1
    for j in lst1:
        if j in dic:
           result.append(j)

    return result
            
```

```python
lst1=[4,5,6,7,8]
lst2=[5,6,9]
common_ele(lst1,lst2)
```

```python
def slight_change_above_question(lst,target):
    small={}
    for i in range(len(lst)):
        small[lst[i]]=i
    for j in range(len(lst)):
        result=target-lst[i]
        if result in small:
            return i,small[result]
```

```python
lst=[2,3,4,7]
slight_change_above_question(lst,5)
```

```python
def element_frequency(lst):
    result={}
    for i in range(len(lst)):
        count=0
        for j in range(len(lst)):
            if lst[i]==lst[j]:
                count+=1
        result[lst[i]]=count
    return result
```

```python
lst=["this","is","a","python","classess","python"]
element_frequency(lst)
```
