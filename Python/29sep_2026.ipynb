```python
# wap to find the sum of all elements of matrix
```

```python
def sum_of_matrix(lst):
    sum=0
    for i in range(len(lst)):
        for j in range(len(lst[0])):
            sum+=lst[i][j]

    print(sum)
```

```python
lst= [[1,2,3],[4,5,6],[7,8,9]]
sum_of_matrix(lst)
```

```python
# WAP to get a product of all element of matrix
```

```python
def product(lst):
    product=1
    for i in range(len(lst)):
        for j in range(len(lst[0])):
            product*=lst[i][j]

    return product
```

```python
lst= [[1,2,3],[4,5,6],[7,8,9]]
product(lst)
```

```python
# WAP to cheak a particular element is persent in the matrix or not
```

```python
def chek_element(lst,element):
    for i in range(len(lst)):
        for j in range(len(lst[0])):
            if element==lst[i][j]:
                return True
    else:
        return False
```

```python
lst= [[1,2,3],[4,5,6],[7,8,9]]
chek_element(lst,11)
```

```python
# WAP to cheak a particular matrix is a diagonal matrix or not 
```

```python
def chek_element(lst):
    for i in range(len(lst)):
        for j in range(len(lst[0])):
            if lst[i][j]!=0 and i!=j :
                return False
    else:
        return True
```

```python
lst= [[1,0,0],[0,5,0],[0,0,9]]
chek_element(lst)
```

```python
def trace(lst):
    tr=0
    for i in range(len(lst)):
        tr+=lst[i][i]

    return tr
```

```python
lst= [[1,0,0],[0,5,0],[0,0,9]]
trace(lst)
```

```python
def transpose(matrix):
    for i in range(len(matrix)):
        for j in range(i+1,len(matrix)):
            matrix[i][j],matrix[j][i]=matrix[j][i],matrix[i][j]
    return matrix
            
```

```python
matrix=[[1,2,3],[4,5,6],[7,8,9]]
transpose(matrix)
```
