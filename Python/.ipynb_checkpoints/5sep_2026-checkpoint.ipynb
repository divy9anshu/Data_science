# WAP a program to get the sum of all even number of n number

```python
n = int(input("Entr any number"))

i=1
even_sum=0
while i<=n:
    if i%2==0:
        even_sum+=i
    i+=1
print(even_sum)

        
```

# WAP to get difference betwn sum of even and sum of odd

```python
n = int(input("Entr any number"))
i=1
even_sum=0
odd_sum=0
while i<=n:
    if i%2==0:
        even_sum+=i
    else:
        odd_sum+=i
    i+=1
print(even_sum-odd_sum)
```

# WAP to find the factorial of a number 

```python
n = int(input("Enter the number"))

fact =1
i=1
while i<=n:
    fact*=i
    i+=1
print(fact)
```
