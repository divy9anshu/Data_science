11. Quadratic Equation Roots
Write a Python code to calculate the root of a quadratic equation.
Test Data : 1 5 7
Expected Output :
Root are imaginary;
No solution.

```python
a = int(input("Enter the number a"))
b = int(input("Enter the number b"))
c = int(input("Enter the number c"))

D = ((b**2)-(4*a*c))

if D > 0:
    print("Two real and different roots")
elif D < 0:
    print("Roots are imaginary/complex")
else:
    print("Two real and equal roots")
```

customer.

Unit Charge/unit

upto 199 @1.20

200 and above but less than 400 @1.50

If bill exceeds Rs. 400 then a surcharge of 15% will be charged and the
minimum bill should be of Rs. 100/-

James
800

Customer IDNO :1001
Customer Name :James
unit Consumed :800
Amount Charges @Rs. 2.00 per unit : 1600.00

25. Geometrical Shapes Area Calculator 
Write a Python code which computes the area of various geometrical shapes 
using a menu-driven approach. 
Test Data : 
1 
5 
Expected Output : 
The area is : 78.500000 fomula 

```python
import math

print("1. Circle")
print("2. Rectangle")
print("3. Triangle")
print("4. Square")

choice = int(input("Enter your choice: "))

if choice == 1:
    radius = float(input("Enter the radius: "))
    area = math.pi * radius * radius
    print("The area is :", area)

elif choice == 2:
    length = float(input("Enter the length: "))
    breadth = float(input("Enter the breadth: "))
    area = length * breadth
    print("The area is :", area)

elif choice == 3:
    base = float(input("Enter the base: "))
    height = float(input("Enter the height: "))
    area = 0.5 * base * height
    print("The area is :", area)

elif choice == 4:
    side = float(input("Enter the side: "))
    area = side * side
    print("The area is :", area)

else:
    print("Invalid choice")
```
