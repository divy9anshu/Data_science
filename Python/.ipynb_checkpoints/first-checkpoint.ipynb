```python

radius=float(input("Enter a radius of circle"))
x_cor= float(input("Enter the number x"))
y_cor= float(input("Enter the number y"))

distance = (((x_cor)**2)+((y_cor)**2))**0.5

if distance == radius:
    print("point lies on the cirle")
if distance > radius:
    print("point lies outside the cirle")
if distance <radius:
    print("point lies inside the circle")
    
```
