```python
def set_add(s,element):
    lst=list(s)
    result=lst+[element]
    return set(result)
```

```python
s={2, 3, 4, 5, 6, 7, 8, 45, 5000}
set_add(s,50)
```

```python
def set_remove(s,el):
    lst1=list(s)
    result=[]
    for i in lst1:
        if i!=el:
            result.append(i)
    return set(result)
```

```python
s={67,89,100,300,400}
el=400
set_remove(s,el)
```

```python
def set_union(s1,s2):
    result=[]
    result=list(s1)+list(s2)
    return set(result)
```

```python
def set_intersection(s1,s2):
    result=set()
    for i in s1:
        for j in s2:
            if i==j:
                result.add(i)
    return result
```

```python
def set_intersection(s1,s2):
    result=set()
    for i in s1:
        if i not in s2:
            result.add(i)
    return result
```

```python
s1={1,2,3,4}
s2={4,7,8,9}
set_intersection(s1,s2)
```

```python
def symetric_difference(s1,s2):
    result1=[]
    result2=[]
    result=[]
    lst1=list(s1)
    lst2=list(s2)
    for i in lst1:
        if i not in lst2:
            result1.append(i)
    for j in lst2:
        if j not in lst1:
            result2.append(j)
    result=result1+result2
    return result
```

```python
s1={1,2,3,4}
s2={4,7,8,9}
symetric_difference(s1,s2)
```
