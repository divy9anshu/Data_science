# WAP using function 
n=5

```python
def sos_using_function(n):
    i = 1
    sum_r = 0
    while i <= n:
        value = square(i)
        sum_r = sos(sum_r, value)
        i += 1
    print(sum_r)
def square(x):
    value = x ** 2
    return value
def sos(sum_r, value):
    sum_r += value
    return sum_r
```

```python
sos_using_function(5)
```

```python
def avg(n):
    sum=0
    for i in range(1,n+1):
        sum+=i

    return sum/n
def input_passing(n):
    avrage=avg(n)
    i=1
    transformation=0
    for i in range(n+1):
        transformation=((i-avrage)/2.87)

    print(transformation)
```
