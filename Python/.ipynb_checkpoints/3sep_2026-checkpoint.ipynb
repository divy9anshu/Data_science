```python
a = int(input("Enter any number "))

if a%5==0:
    if a%10==0:
        print("hello")
    else:
        print("bye")
```

```python
a = int(input("Enter any number "))

if a%5==0:
    if a%10==0:
        print("hello")
    else:
        print("bye")
else:
    print("bbye")
```

WAP to cheak a particular year is lear year or not.

```python
year = int(input("Enter any year: "))

if year % 400 == 0:
    print("Leap year")
else:
    if year % 100 == 0:
        print("Not a leap year")
    else:
        if year % 4 == 0:
            print("Leap year")
        else:
            print("Not a leap year")
```
