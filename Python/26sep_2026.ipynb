```python
def my_remove(lst, element):
    result = []
    
    for i in lst:
        if i != element:
            result.append(i)
    
    return result


lst = [10, 20, 30, 20, 40]
element = 20

print(my_remove(lst, element))
```

```python
lst1=[3,4,5,6,7,8,9]
lst2=[5,6,7,10,15]
```

```python
def commonelment(lst1,lst2):
    result=[]
    for i in lst1:
        for j in lst2:
            if i==j:
                result.append(i)

    return result
            
```

```python
lst1=[3,4,5,6,7,8,9]
lst2=[5,6,7,10,15]
commonelment(lst1,lst2)
```

```python
def pop_fun(lst,index):
    result=[]
    for i in range(len(lst):
        if i != index:
            result.append(lst[i])
    return lst[index],result
    
        
```

```python
lst=[4,5,6,7,8,9]
pop_fun(lst,2)
```

```python
lst=[2,2,2,2,3,4,4,5,6,7,9,7,9]
```

```python
def odd_element(lst):
    ans=0
    for i in lst:
        ans^=i
    return ans
```

```python
lst=[2,2,2,2,3,4,4,5,6,7,9,7,9]
odd_element(lst)
```
