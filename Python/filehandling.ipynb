```python
# WAP to count the number of words in file
f2=open("myfile.txt","r")
data=f2.read()
lst=data.split()
print(len(lst))
f2.close()
```

```python
# Wap to count the number of vowels and consonants in file
f3=open("myfile.txt","r")
data=f3.read()
vowel_count=0  
consonanat_count=0
for ch in data:
    if ch in ["a","e","i","o","u"]:
        vowel_count+=1
    else:
        consonanat_count+=1
print(vowel_count,consonanat_count)
f3.close()
```
