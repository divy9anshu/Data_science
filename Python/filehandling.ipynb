# WAP to find the count of vowel and consontent in each line and paste the count to othere file

```python
f1=open("file1.txt","r")
f2=open("result.txt","w")
lst=f1.readlines()
vowel_count=0
consonant_count=0
for line in lst:
    for ch in line:
        if ch in ["a","e","i","o","u"]:
            vowel_count+=1
        elif ch.isspace()==False and ch not in ["a","e","i","o","u"]:
            consonant_count+=1
f2.write(str(consonant_count))
f2.write("\n")
f2.write(str(vowel_count))
f1.close()
f2.close()
```
