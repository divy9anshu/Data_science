```python
def recursive_multiplication(a,b):
    if b==0:
        return 0
    else:
        return a+recursive_multiplication(a,b-1)
```

```python
recursive_multiplication(6,4)
```

```python
recursive_multiplication(9,3)
```

# WAP to perform a division using recutsion:


```python
def recursive_division(a,b):
    if a==0:
        return 0
    else:
        return 1+recursive_division((a-b),b)
```

```python
recursive_division(6,2)
```

```python
def fun1(a,n):
    if n==0:
        return 1
    else:
        return a*fun1(a,n-1)
```

```python
fun1(2,8)
```

```python
def foo(n):
    if n==0:
        return 1
    else:
        return n+foo(n-1)
        
```
