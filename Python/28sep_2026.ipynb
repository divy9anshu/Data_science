```python
def reverse_fun(lst):
    i=0
    j=len(lst)-1
    while i<j :
        lst[i],lst[j]=lst[j],lst[i]
        i+=1
        j-=1
    return lst
        
```

```python
lst=[45,12,34,78,17,9,5]
reverse_fun(lst)
```

```python
def sort_lst(lst):
    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            if lst[i]>lst[j]:
                lst[i],lst[j]=lst[j],lst[i]
    return lst
```

```python
lst=[45,12,34,78,17,9,5]
sort_lst(lst)
```

# 2. WAP to find the 2nd largest element in list

```python
def second_largest(lst):
    largest = lst[0]
    second = lst[0]

    for i in range(len(lst)):
        if lst[i] > largest:
            second = largest
            largest = lst[i]
        elif lst[i] > second and lst[i] != largest:
            second = lst[i]

    return second

def largest_fun(lst):
    for i in range(len(lst)):
        for j in range(i+1,len(lst)):
            if lst[i]>lst[j]:
                return lst[i]
            
                
```

```python
lst=[45,12,34,78,17,9,5]
largest_fun(lst)
second_largest(lst)
```
