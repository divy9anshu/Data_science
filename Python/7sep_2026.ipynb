# WAP to check a number is perfect number or not 

```python
n = int(input("Enter any number "))
i=1
sum=0
while i<n:
    if n%i==0:
        sum+=i
    i+=1

if i==sum:
    print("perfect Number")
else:
    print("not perfect Number")
```
