# Q1. The Dragon&#39;s Coins (For Loop) Story: A brave knight enters a dragon&#39;s cave where there are 100 treasure chests. Every 10th chest contains 10 gold coins. Others are empty. Task: Write a program to calculate how many coins the knight collects. Input: Number of chests n = 100 Output: Total coins collected. give the approch

```python
n = 100
total = 0

for i in range (1,n):
    if i % 10 == 0:
        total = total + 10

print(total)
```
