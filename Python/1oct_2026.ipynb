```python
def recursive_length(t1):
    if t1==():
        return 0
    return 1+recursive_length(t1[1:])
```

```python
def count_tuple(t2,element):
    if t2==():
        return 0
    elif t2[0]==element:
        return 1+count_tuple(t2[1:],element)
    else:
        return count_tuple(t2[1:],element)
    
```

```python
t2=(1,2,4,5,6,6,4,5,6,4,5,6,8,5,5,4,4)
count_tuple(t2,4)
```

```python
def recursive_sum(t1):
    
    if t1==():
        return 0
    else:
        return t1[0]+recursive_length(t1[1:])
```

```python
t1=(1,2,4,5,6,6)
recursive_sum(t1)
```

```python
def sum_of_lists(lst):
    if lst==[]:
        return 0
    elif type(lst[0])==list:
        if lst[0]==[]:
            return 0
        else:
            return lst[0][0]+sum_of_lists(lst[1:][0:])
    else:
        return lst[0]+sum_of_lists(lst[1:])
    
```

```python
lst=[2,3,4,[5,6],[6,7],100]
sum_of_lists(lst)
```
