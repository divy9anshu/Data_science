# while loop

```python
i=1 
while i <=10:
    if i !=3:
        print(i+2>>1)
    else:
        print(i<<2)
    i*=2
```

```python
i= 2 
while i<=10:
    if i&1!=0:
        print(i*2<<1)
    else:
        print(i*3<<1)
    i<<=1
```

# wap n number in a reverse

```python
n = int(input("Enter the n-value: "))

i = n

while i >= 1:
    print(i)
    i -= 1
```
