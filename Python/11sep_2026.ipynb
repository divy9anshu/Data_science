```python
def count_no_set_bits(number):
    count = 0
    while number>0:
        if number&1==1:
            count+=1
        number>>=1
    return count
    
```

```python
number = int(input("Enter any number "))
count_no_set_bits(number)
```

```python
def factorial_sod(number):
    sum_d=0
    while number>0:
        digit=number %10
        sum_d+=digit
        number//=10
    ans=factorial(sum_d)
    return ans
def factorial(n):
    i=1
    fact=1
    while i<=n:
        fact*=i
        i+=1
    return fact
```

```python
def printing_all_digit(string):
    for i in range(len(string)):
        for j in range(i+1,len(string)+1):
            ss=string[i:j]
            if is_palindrom(ss)==True:
                print(ss)

def is_palindrom(ss):
    i=0
    j=len(ss)-1
    while i<=j:
        if ss[i]!=ss[j]:
            return False
        else:
            return True
```

```python
string ="12345678"
printing_all_digit(string)
```
