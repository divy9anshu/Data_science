# WAP to find the sum of factorial of each digit.

```python
n = int(input("Enter a number n"))
fact_sum=0
while n>0:
    digit =n%10
    i=1
    fact=1
    while i<=digit:
        fact*=i
        i+=1
    fact_sum+=fact
    n//=10
print(fact_sum)
```

# WAP to get a square of each ddigit and find a factorial of each difigt and sum it 

```python
n = int(input("Enter a number n"))
fact_sum=0
digit_square=0
while n>0:
    digit =n%10
    digit_square=digit**2
    i=1
    fact=1
    while i<=digit_square:
        fact*=i
        i+=1
    fact_sum+=fact
    n//=10
print(fact_sum)
```

# WAP to print all the pythagoarus triplets in n using for loop

```python
n = int(input("Enter the value of n: "))

for i in range(0, n + 1,1):
    for j in range(i+ 1, n + 1,1):
        for k in range(j + 1, n + 1,1):
            if i*i + j*j == k*k:
                print(i, j, k)
```
