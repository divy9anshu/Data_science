```python
def print_n(n):
    if n==0:
        return
    else:
        
        print_n(n-1)
        print(n)
```

```python
print_n(10)
```

```python
def product(n):
    if n==1:
        return 1
    else:
        return n*product(n-1)
```

```python
# WAP to find the power of a number using recursion
```

```python
def power_r(a,n):
    if n==1:
        return 1
    else:
        return a*power_r(a,n-1)
```

```python
power_r(2,5)
```

```python
# WAP to find sum of digit using recursion

```

```python
def sum_digit(n):
    if n==0:
        return n
    else:
        digit=n%10
        
        return digit+sum(n//10)
```

```python
sum_digit(123)
```
