# expception handling
    -handing the runtime error
    -execptional handling is not about rectifying the runtime error its all about finding the alternate way to do that thir
    so thet ther will not be any kind of abnormal termination.
    -we are having a exceptional handling blocks:
    1.try:
    -it is having a risky code.
    -we will keep theat cide in try which can raise the exception
    
    2.except:
    -handling code
    -if theexception is getting raise this handling block will handle the exception
    3.finally:
    -it is just for cleanup/refreshing purpose
    -weather the exeption raise or not the finally block is getting executed.
    

```python
try:
    print(10/0)
except:
    print("cannot divide by zero")
```
